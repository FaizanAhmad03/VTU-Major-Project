```python
# ============================================================
# AQI PREDICTION USING MACHINE LEARNING
# Dataset: City, Date, Pollutants, AQI, AQI_Bucket
# Model: Random Forest Regression
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import joblib

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# IMPORTANT:
# If 'df' is already loaded from your previous cell,
# this will use that dataset.
#
# If df does not exist, upload the ZIP file again.

try:
    df
    print("Using the dataset already loaded in Colab.")

except NameError:

    from google.colab import files
    import zipfile
    import io
    import os

    print("Please upload your ZIP dataset.")

    uploaded = files.upload()

    zip_filename = list(uploaded.keys())[0]

    with zipfile.ZipFile(
        io.BytesIO(uploaded[zip_filename]), 'r'
    ) as z:

        csv_files = [
            f for f in z.namelist()
            if f.lower().endswith('.csv')
        ]

        if not csv_files:
            raise ValueError("No CSV file found inside ZIP.")

        csv_filename = csv_files[0]

        try:
            with z.open(csv_filename) as f:
                df = pd.read_csv(f, encoding='utf-8')

        except UnicodeDecodeError:

            with z.open(csv_filename) as f:
                df = pd.read_csv(f, encoding='latin1')


print("\nDataset loaded successfully!")


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\nDataset shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 4. CLEAN COLUMN NAMES
# ============================================================

# Remove accidental spaces from column names
df.columns = df.columns.str.strip()

print("\nCleaned column names:")
print(df.columns.tolist())


# ============================================================
# 5. CONVERT DATE COLUMN
# ============================================================

df['Date'] = pd.to_datetime(
    df['Date'],
    errors='coerce'
)

# Extract useful time information
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day


# ============================================================
# 6. CHECK MISSING VALUES
# ============================================================

print("\nMissing values before cleaning:")
print(df.isnull().sum())


# ============================================================
# 7. SELECT FEATURES
# ============================================================

# These are the pollutant measurements we will use
features = [
    'PM2.5',
    'PM10',
    'NO',
    'NO2',
    'NOx',
    'NH3',
    'CO',
    'SO2',
    'O3',
    'Benzene',
    'Toluene',
    'Xylene'
]

target = 'AQI'


# Check that all required columns exist
missing_features = [
    col for col in features + [target]
    if col not in df.columns
]

if missing_features:
    raise ValueError(
        f"These columns are missing: {missing_features}"
    )


# ============================================================
# 8. CONVERT FEATURES TO NUMERIC
# ============================================================

for col in features + [target]:
    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )


# ============================================================
# 9. REMOVE ROWS WHERE AQI IS MISSING
# ============================================================

# We cannot train a supervised model if the target AQI
# is missing.

df = df.dropna(
    subset=[target]
)

print("\nDataset shape after removing missing AQI:")
print(df.shape)


# ============================================================
# 10. HANDLE MISSING POLLUTANT VALUES
# ============================================================

# Fill missing pollutant values using the median
# of each pollutant.

for col in features:
    df[col] = df[col].fillna(
        df[col].median()
    )


print("\nMissing values after cleaning:")
print(df[features + [target]].isnull().sum())


# ============================================================
# 11. CREATE X AND y
# ============================================================

X = df[features]

y = df[target]

print("\nInput features:")
print(X.columns.tolist())

print("\nTarget:")
print("AQI")


# ============================================================
# 12. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ============================================================
# 13. CREATE RANDOM FOREST MODEL
# ============================================================

model = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

print("\nTraining Random Forest model...")

model.fit(
    X_train,
    y_train
)

print("Model training completed!")


# ============================================================
# 14. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(X_test)


# ============================================================
# 15. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n================================")
print("MODEL PERFORMANCE")
print("================================")

print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")


# ============================================================
# 16. ACTUAL VS PREDICTED TABLE
# ============================================================

results = pd.DataFrame({
    'Actual AQI': y_test.values,
    'Predicted AQI': y_pred
})

results['Error'] = (
    results['Actual AQI']
    - results['Predicted AQI']
)

print("\nFirst 20 predictions:")

display(
    results.head(20)
)


# ============================================================
# 17. ACTUAL VS PREDICTED GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

# Perfect prediction line

min_value = min(
    y_test.min(),
    y_pred.min()
)

max_value = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle='--'
)

plt.xlabel("Actual AQI")
plt.ylabel("Predicted AQI")

plt.title(
    "Actual AQI vs Predicted AQI"
)

plt.show()


# ============================================================
# 18. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({
    'Feature': features,
    'Importance': model.feature_importances_
})

importance = importance.sort_values(
    by='Importance',
    ascending=False
)

print("\nFeature importance:")

display(importance)


# ============================================================
# 19. FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.barh(
    importance['Feature'],
    importance['Importance']
)

plt.xlabel("Importance")
plt.ylabel("Pollutant")

plt.title(
    "Feature Importance for AQI Prediction"
)

plt.gca().invert_yaxis()

plt.show()


# ============================================================
# 20. SAVE TRAINED MODEL
# ============================================================

model_filename = "AQI_RandomForest_Model.pkl"

joblib.dump(
    model,
    model_filename
)

print("\nModel saved as:")
print(model_filename)


# ============================================================
# 21. EXAMPLE: PREDICT AQI FROM NEW POLLUTANT VALUES
# ============================================================

new_data = pd.DataFrame({

    'PM2.5': [50],
    'PM10': [80],
    'NO': [10],
    'NO2': [30],
    'NOx': [40],
    'NH3': [20],
    'CO': [1.0],
    'SO2': [15],
    'O3': [50],
    'Benzene': [2],
    'Toluene': [5],
    'Xylene': [1]

})


predicted_aqi = model.predict(
    new_data
)

print("\n================================")
print("NEW AQI PREDICTION")
print("================================")

print(
    f"Predicted AQI: {predicted_aqi[0]:.2f}"
)


# ============================================================
# 22. DOWNLOAD MODEL
# ============================================================

from google.colab import files

files.download(
    model_filename
)
```
